# 01 - Data Loading

This notebook loads the raw Telco Customer Churn CSV from `../data/`, inspects its shape, dtypes, missing values, and summary statistics, and saves the raw DataFrame as a pickle for the next notebook (`02_eda.ipynb`) to consume.

## 1. Imports

```python
import pandas as pd
import numpy as np
import glob
import os

pd.set_option('display.max_columns', None)
```

## 2. Locate and load the CSV

We look for any CSV file in `../data/` so this notebook works regardless of the exact filename of the dataset you place there (e.g. `WA_Fn-UseC_-Telco-Customer-Churn.csv`).

```python
data_dir = '../data'
csv_candidates = glob.glob(os.path.join(data_dir, '*.csv'))

if not csv_candidates:
    raise FileNotFoundError(
        f'No CSV file found in {data_dir}. Place the Telco Customer '
        'Churn CSV there before running this notebook.'
    )

csv_path = csv_candidates[0]
print(f'Loading: {csv_path}')
df = pd.read_csv(csv_path)
df.head()
```

## 3. Shape

Number of rows (customers) and columns (features).

```python
print(f'Shape: {df.shape}')
print(f'Rows: {df.shape[0]}, Columns: {df.shape[1]}')
```

## 4. Data types

Note that `TotalCharges` is often read in as an `object` (string) column even though it is numeric, because a handful of rows contain blank strings instead of numbers. We will handle this in the preprocessing notebook.

```python
df.dtypes
```

## 5. Missing values

A per-column count of missing (NaN) values. Note this will **not** catch the blank-string issue in `TotalCharges`, since those are '' strings rather than true NaNs at this stage.

```python
df.isnull().sum()
```

## 6. `df.info()`

A concise summary of columns, non-null counts, and dtypes.

```python
df.info()
```

## 7. `df.describe()`

Summary statistics for the numeric columns, plus a categorical summary for completeness.

```python
df.describe()
```

```python
df.describe(include='object')
```

## 8. Target distribution

A quick look at the class balance of the `Churn` target.

```python
df['Churn'].value_counts()
df['Churn'].value_counts(normalize=True)
```

## 9. Save the raw DataFrame

We pickle the raw, unmodified DataFrame to `../data/raw_df.pkl` so that every downstream notebook (EDA, preprocessing, ...) starts from the exact same raw data without needing to re-parse the CSV or re-discover its filename.

```python
raw_path = os.path.join(data_dir, 'raw_df.pkl')
df.to_pickle(raw_path)
print(f'Saved raw DataFrame to {raw_path}')
```
