# 04 - Model Training

This notebook loads the preprocessed, SMOTE-resampled train/test pickles and trains two classifiers: a Logistic Regression baseline and an XGBoost classifier. Both trained models are saved to `../models/` for evaluation in `05_evaluate.ipynb`.

## 1. Imports and load data

```python
import pandas as pd
import numpy as np
import joblib
import os
from sklearn.linear_model import LogisticRegression
from xgboost import XGBClassifier

X_train = pd.read_pickle('../data/X_train.pkl')
X_test = pd.read_pickle('../data/X_test.pkl')
y_train = pd.read_pickle('../data/y_train.pkl')
y_test = pd.read_pickle('../data/y_test.pkl')

print(f'X_train: {X_train.shape}, X_test: {X_test.shape}')
os.makedirs('../models', exist_ok=True)
```

## 2. Logistic Regression baseline

Logistic Regression is a simple, fast, and highly interpretable linear model — a natural first baseline for a binary classification problem like churn prediction. Its coefficients are also easy to explain to a business stakeholder (e.g. "month-to-month contracts increase churn odds").

```python
log_reg = LogisticRegression(max_iter=1000, random_state=42)
log_reg.fit(X_train, y_train)
print('Logistic Regression trained.')
print(f'Train accuracy: {log_reg.score(X_train, y_train):.4f}')
print(f'Test accuracy:  {log_reg.score(X_test, y_test):.4f}')
```

```python
joblib.dump(log_reg, '../models/logistic_regression.pkl')
print('Saved Logistic Regression model to ../models/logistic_regression.pkl')
```

## 3. XGBoost classifier

XGBoost is a gradient-boosted tree ensemble that typically outperforms linear models on tabular data with non-linear feature interactions (e.g. how tenure, contract type, and monthly charges interact to drive churn), at the cost of some interpretability. We use it here as a stronger candidate model to compare against the Logistic Regression baseline.

```python
xgb_clf = XGBClassifier(
    n_estimators=200,
    max_depth=4,
    learning_rate=0.1,
    subsample=0.8,
    colsample_bytree=0.8,
    eval_metric='logloss',
    random_state=42,
)
xgb_clf.fit(X_train, y_train)
print('XGBoost trained.')
print(f'Train accuracy: {xgb_clf.score(X_train, y_train):.4f}')
print(f'Test accuracy:  {xgb_clf.score(X_test, y_test):.4f}')
```

```python
joblib.dump(xgb_clf, '../models/xgboost_model.pkl')
print('Saved XGBoost model to ../models/xgboost_model.pkl')
```
